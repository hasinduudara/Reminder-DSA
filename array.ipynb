```python
# Q4
arr = [5, 10, 15, 20]

def check_number(arr):
    target = int(input("Enter number: "))
    if target in arr:
        print("Found")
    else:
        print("Not found")

check_number(arr)
```

```python
# Q5

arr = [1, 2, 3, 4, 5]
ra = []

for i in range(len(arr) - 1, -1, -1):
    ra.append(arr[i])

print(ra)
```

```python
# Q6

arr = [15, 8, 42, 23, 7]

min = arr[0]

for i in range(1, len(arr)):
    if arr[i] < min:
        min = arr[i]

print(arr[i])
```
