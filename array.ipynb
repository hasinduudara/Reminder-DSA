```python
num = [10, 20, 30, 40]

l = len(num)

for i in range(l):
    print(num[i])
```

```python
arr = [15, 8, 42, 23, 7]

max  = arr[0]

for i in range(1, len(arr)):
    if arr[i] > max:
        max = arr[i]

print(max)
```
