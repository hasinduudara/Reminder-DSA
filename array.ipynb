```python
# Q8

arr = [1, 3, 10, 7, 9]
is_sorted = True

for i in range(len(arr) - 1):
    if arr[i] > arr[i +1]:
        is_sorted = False
        break

if is_sorted:
    print("sorted")
else:
    print("not sorted")
```

```python
arr = [12, 35, 1, 10, 34, 1]

largest = 0
second_largest = 0

for num in arr:
    if num > largest:
        second_largest = largest
        largest = num
    elif num > second_largest and num != largest:
        second_largest = num

print(second_largest)
```
